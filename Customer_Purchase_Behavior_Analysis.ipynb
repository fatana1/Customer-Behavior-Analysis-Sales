# Load the data

```python
import pandas as pd

df = pd.read_csv('C:/Users/lab\Desktop/Data_analysis_Project/PBI_2026/customer-trends-data-analysis-SQL-Python-PowerBI-main/customer_shopping_behavior.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
# Summary statistics using .describe()
df.describe(include='all')
```

```python
# Checking if missing data or null values are present in the dataset

df.isnull().sum()
```

```python
# Imputing missing values in Review Rating column (37) with the median rating of the product category

df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
# Renaming columns according to snake casing for better readability and documentation

df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
df.head(5)
```

```python
# create a new column age_group
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior'] # since age is in number we make new column as young adult..
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

```python
labels
```

```python
df[['age','age_group']].head(10)
```

```python
# create new column purchase_frequency_days, we create a dictionay to create a new column 

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df[['discount_applied','promo_code_used']].head(10)
```

```python
(df['discount_applied'] == df['promo_code_used']).all()
# both of this culumn carry the same information so we need to delet one 
```

```python
# Dropping promo code used column

df = df.drop('promo_code_used', axis=1) # axis =1 means that the column should remove not row 
```
